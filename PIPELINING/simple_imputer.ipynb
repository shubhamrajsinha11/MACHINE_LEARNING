```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, RobustScaler
from category_encoders import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression 
from sklearn.model_selection import train_test_split
```

```python
data2= {
    'emp_name': ('shubham','karan',np.nan,'aadi',np.nan),
    'dname':('clerk',np.nan,'accounts',np.nan,'clerk'),
    'comm':(1000,np.nan,np.nan,90000,200),
    'dno': (100,np.nan,20,20,np.nan),
    'sal': (10000,1500000,np.nan,8000,800000)
    
}
```

```python
df1=pd.DataFrame(data2)
df1
```

```python
imputer6= SimpleImputer(strategy='median')
imputer6.fit_transform(df1[['sal','comm','dno']])
```

```python
numerical_cols = Pipeline(
    steps=[
        ('Imputer',SimpleImputer(strategy='median')),
        ('scaling',RobustScaler())
    ])
```

```python
categorical_cols = Pipeline(
    steps=[
        ('Imputer',SimpleImputer(strategy='median')),
        ('scaling',OneHotEncoder(handle_unknown='ignore'))
    ])
```

```python
preprocessing= ColumnTransformer(
    transformers=[
        ('numerical_pipeline',numerical_cols,['sal','comm','dno']),
        ('categorical_pipeline',categorical_cols,['emp_name','dname'])
    ]
)
```

```python
main_pipeline=Pipeline(
    steps=[
        ('pre',preprocessing),
        ('model', LinearRegression())
    ]
)
```

```python
X=df.drop(columns=['sal'])
Y=df['sal']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(X,Y,train_size=0.8,random_state=42)
```

```python
main_pipeline.fit(xtrain,ytrain)
```
