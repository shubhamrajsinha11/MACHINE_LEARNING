```python
import pandas as pd 
from sklearn.preprocessing import RobustScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import matplotlib as plt
import seaborn as sns 
```

```python
df=pd.read_csv('loan_data.csv')
df
```

```python
x=df.drop(columns=['loan_status'])
y=df.loan_status
```

```python
xtrain,xtest,ytrain,ytest= train_test_split(x,y, trai_size=0.8, random_state=42)
```

```python
num_cols=x.select_dtypes(include='number').columns
obj_cols=x.select_dtypes(include='object').columns
```

```python
preprocessing=ColumnTransformer(
    transformers=[
        ('scaling'.RobustScaler(),num_cols),
        ('encoder',OneHotEncoder(handle_unknown='ignore'),obj_cols)
    ]
)
```

```python
from sklearn.linear_model import LinearRegression


main_pipeline= Pipeline(
    steps=[
        ('pre',preprocessing),
        ('model',LinearRegression(penalty='l2', solver='saga', class_weight='balanced'))
    ]
)
main_pipeline.fit(xtrain,ytrain)
```

# model evaluation

### train data

```python
ytrain_pred=main_pipeline.predict(xtrain)
print(classification_report(ytrain,ytrain_pred))
```

out of all the predicted 0 labels, 93% labels are correctly predicted as 0.
out of all the predicted 1 labels, 73% labels are correctly predicted as 1.

```python
ytest_pred=main_pipeline.predict(xtest)
print(classification_report(ytest,ytest_pred))
```
