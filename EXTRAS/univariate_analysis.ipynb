```python
import pandas as pd
```

```python
df=pd.read_csv('Titanic.csv')
```

```python
df.head()
```

### working with categorical data 

```python
# countplot
import seaborn as sns
```

```python
sns.countplot(data=df, x='Pclass', hue='Survived')
# df['Survived'].value_counts().plot(kind='bar', title='Survived Count')
```

```python
# piechart
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f', title='Survived Distribution')
```

### working with numerical data

```python
import matplotlib.pyplot as plt
```

```python
# histogram
plt.hist(df['Age'])
```

```python
# distplot
sns.distplot(df['Age'])
# the curve in this plot is called kernek density estimation (KDE) which is a way to estimate the probability density function of a continuous random variable. It is a smoothed version of the histogram and provides a better understanding of the distribution of the data.
# PDF-> this is the probability density function which is a function that describes the likelihood of a random variable taking on a particular value. It is used to model continuous random variables and is often used in statistical analysis and machine learning.
```
