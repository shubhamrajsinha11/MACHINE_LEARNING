```python
import pandas as pd
from sklearn.ensemble import IsolationForest
```

```python
df=pd.read_csv('creditcard.csv')
df.head()
```

```python
df.shape
```

```python
model=IsolationForest(n_estimators=100,
                      max_samples=0.8,
                      max_features=1,
                      bootstrap=False,
                      n_jobs=1,
                      verbose=5,
                      random_state=42,
                      contamination=0.05 #5% #auto
                      )
result=model.fit_predict(df)
anomaloies=df[result==-1]
normal=df[result==1]
```
